```python
%pip install -qq seaborn
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
measurements = {"Dopamine Level (ng/mL)": [45, 50, 42, 55, 48, 53, 46, 51],
                 "Reward Score (0-100)": [70, 80, 65, 85, 75, 82, 72, 78]}
measurement_df = pd.DataFrame(data = measurements)

plot = sns.lmplot(measurement_df, x = "Dopamine Level (ng/mL)", y = "Reward Score (0-100)")
plot.set(title = "Dopamine vs Reward Score")
```

```python
import numpy as np

x = measurement_df["Dopamine Level (ng/mL)"]
y = measurement_df["Reward Score (0-100)"]

slope, intercept = np.polyfit(x, y, 1)

print("Slope:", slope)
print("Intercept:", intercept)
y_pred = slope * x + intercept
ss_res = np.sum((y - y_pred) ** 2)        # residual sum of squares
ss_tot = np.sum((y - np.mean(y)) ** 2)   # total sum of squares
r_squared = 1 - (ss_res / ss_tot)

print("R^2 =", r_squared)
```
